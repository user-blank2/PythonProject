```python
# Load dataset — one line!
dataset = load_dataset("dair-ai/emotion")
print(dataset)
print("\nSample entry:")
print(dataset["train"][0])
```

```python
# Emotion label map
id2label = {0: "sadness", 1: "joy", 2: "love", 3: "anger", 4: "fear", 5: "surprise"}
label2id = {v: k for k, v in id2label.items()}

# Quick look at class distribution
from collections import Counter
labels = dataset["train"]["label"]
counts = Counter(labels)
print("\nClass distribution in training set:")
for label_id, count in sorted(counts.items()):
    print(f"  {id2label[label_id]:10s} ({label_id}): {count} samples")
```

## Step 4 — Tokenize the Data


```python
MODEL_NAME = "distilbert-base-uncased"
tokenizer = AutoTokenizer.from_pretrained(MODEL_NAME)

# Show tokenization example — great talking point!
example = "I feel so happy today!"
tokens = tokenizer.tokenize(example)
token_ids = tokenizer.encode(example)
print(f"Text    : {example}")
print(f"Tokens  : {tokens}")
print(f"Token IDs: {token_ids}")
```

```python
def tokenize_fn(batch):
    return tokenizer(batch["text"], truncation=True, max_length=128)

tokenized_dataset = dataset.map(tokenize_fn, batched=True)
print("Tokenization complete!")
print("Columns:", tokenized_dataset["train"].column_names)
```

## Step 5 — Load Pre-trained Model

**DistilBERT** is a smaller, faster version of BERT:
- BERT: 110M parameters
- DistilBERT: 66M parameters — 40% smaller, 60% faster, 97% of BERT's accuracy

We add a **classification head** on top (a linear layer) to output 6 emotion probabilities.

```python
model = AutoModelForSequenceClassification.from_pretrained(
    MODEL_NAME,
    num_labels=6,
    id2label=id2label,
    label2id=label2id
)

# Count parameters
total_params = sum(p.numel() for p in model.parameters())
trainable_params = sum(p.numel() for p in model.parameters() if p.requires_grad)
print(f"Total parameters    : {total_params:,}")
print(f"Trainable parameters: {trainable_params:,}")
```

## Step 6 — Define Metrics


```python
accuracy_metric = evaluate.load("accuracy")

def compute_metrics(eval_pred):
    logits, labels = eval_pred
    predictions = np.argmax(logits, axis=-1)
    return accuracy_metric.compute(predictions=predictions, references=labels)
```

## Step 7 — Fine-tune the Model

```python
data_collator = DataCollatorWithPadding(tokenizer=tokenizer)

training_args = TrainingArguments(
    output_dir="./emotion-model",
    num_train_epochs=3,
    per_device_train_batch_size=32,
    per_device_eval_batch_size=32,
    learning_rate=2e-5,
    weight_decay=0.01,
    eval_strategy="epoch",
    save_strategy="epoch",
    load_best_model_at_end=True,
    logging_steps=50,
    report_to="none"   # disable wandb
)

trainer = Trainer(
    model=model,
    args=training_args,
    train_dataset=tokenized_dataset["train"],
    eval_dataset=tokenized_dataset["validation"],
    processing_class=tokenizer,
    data_collator=data_collator,
    compute_metrics=compute_metrics
)

print("Starting fine-tuning...")
trainer.train()
```

## Step 8 — Evaluate on Test Set


```python
results = trainer.evaluate(tokenized_dataset["test"])
print("\n📊 Test Set Results:")
print(f"  Accuracy : {results['eval_accuracy']*100:.2f}%")
print(f"  Loss     : {results['eval_loss']:.4f}")
```

```python
# Detailed classification report
from sklearn.metrics import classification_report

predictions_output = trainer.predict(tokenized_dataset["test"])
preds = np.argmax(predictions_output.predictions, axis=-1)
true_labels = predictions_output.label_ids

print("\n📋 Per-class Classification Report:")
print(classification_report(true_labels, preds, target_names=list(id2label.values())))
```

## Step 9 — Before vs After Comparison

```python
# Determine active device from fine-tuned model
device = next(model.parameters()).device

# Load the BASE model (no fine-tuning, random classification head) and move to the device
base_model = AutoModelForSequenceClassification.from_pretrained(
    MODEL_NAME,
    num_labels=6,
    id2label=id2label,
    label2id=label2id,
    ignore_mismatched_sizes=True
).to(device)

def predict(text, mdl):
    mdl.eval()
    inputs = tokenizer(text, return_tensors="pt", truncation=True, max_length=128)
    # Move inputs to the same device as the model
    inputs = {k: v.to(device) for k, v in inputs.items()}
    with torch.no_grad():
        logits = mdl(**inputs).logits
    probs = torch.softmax(logits, dim=-1)[0]
    pred_id = torch.argmax(probs).item()
    return id2label[pred_id], probs

test_sentences = [
    "I just got promoted at work, I'm so thrilled!",
    "I can't stop crying, everything feels so empty.",
    "This is absolutely outrageous, I'm furious!",
    "I'm scared of what might happen tomorrow.",
    "I love spending time with my family.",
    "Wow, I didn't expect that at all!"
]

print("=" * 65)
print(f"{'Sentence':<40} {'BASE':>10} {'FINE-TUNED':>12}")
print("=" * 65)
for s in test_sentences:
    base_pred, _  = predict(s, base_model)
    ft_pred,   _  = predict(s, model)
    short = s[:38] + ".." if len(s) > 40 else s
    print(f"{short:<40} {base_pred:>10} {ft_pred:>12}")
print("=" * 65)
```

## Step 10 — 🎤 Live Demo Cell


```python
EMOTION_EMOJI = {
    "joy":      "😊",
    "sadness":  "😢",
    "anger":    "😠",
    "fear":     "😨",
    "love":     "❤️",
    "surprise": "😲"
}

def live_predict(text):
    model.eval()
    device = next(model.parameters()).device
    inputs = tokenizer(text, return_tensors="pt", truncation=True, max_length=128)
    # Move inputs to the same device as the model
    inputs = {k: v.to(device) for k, v in inputs.items()}

    with torch.no_grad():
        logits = model(**inputs).logits
    probs = torch.softmax(logits, dim=-1)[0]

    print(f"\n✍️ Input: \"{text}\"")
    print("-" * 45)
    for i, (emotion, prob) in enumerate(zip(id2label.values(), probs)):
        bar = "█" * int(prob.item() * 30)
        marker = " ◀ PREDICTED" if i == torch.argmax(probs).item() else ""
        print(f"  {EMOTION_EMOJI.get(emotion, '❓')} {emotion:8s} {bar:<30} {prob.item()*100:5.1f}%{marker}")
    print()

# ✏️ Change this sentence during your demo!
live_predict("I just got the best news of my life!")
live_predict("I feel completely lost and alone.")
live_predict("How dare they do this to me!")
```
